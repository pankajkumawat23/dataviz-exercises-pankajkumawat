```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# import plotly.express as px

task1_df = df.loc[
    (df["continent"] == "Oceania") & df["year"].isin([2002, 2007])
].copy()
task1_df["year"] = task1_df["year"].astype(str)

fig = px.scatter(
    task1_df,
    x="gdpPercap",
    y="lifeExp",
    color="year",
    hover_name="country",
    color_discrete_map={"2002": "#2878B5", "2007": "#E87500"},
    category_orders={"year": ["2002", "2007"]},
    log_x=True,
    labels={
        "gdpPercap": "GDP per capita (US$; log scale)",
        "lifeExp": "Life expectancy (years)",
        "year": "Year",
    },
    title="Oceania's countries gained income and life expectancy, 2002–2007",
)

# Arrows point from 2002 to 2007 for each country.
for _, country_data in task1_df.groupby("country"):
    before = country_data.loc[country_data["year"] == "2002"].iloc[0]
    after = country_data.loc[country_data["year"] == "2007"].iloc[0]

    fig.add_annotation(
        x=after["gdpPercap"],
        y=after["lifeExp"],
        ax=before["gdpPercap"],
        ay=before["lifeExp"],
        xref="x",
        yref="y",
        axref="x",
        ayref="y",
        showarrow=True,
        arrowhead=2,
        arrowwidth=1,
        arrowcolor="rgba(80, 80, 80, 0.65)",
    )

fig.update_layout(legend_title_text="Year")
fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE
import plotly.express as px

highlight_continent = "Asia"
task2_df = df.loc[df["year"] == 2007].copy()
task2_df["continent_group"] = task2_df["continent"].where(
    task2_df["continent"] == highlight_continent,
    "Other continents",
)

fig = px.scatter(
    task2_df,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    size_max=55,
    color="continent_group",
    hover_name="country",
    hover_data={"continent": True, "pop": ":,", "continent_group": False},
    color_discrete_map={
        highlight_continent: "#2878B5",
        "Other continents": "#B8B8B8",
    },
    log_x=True,
    labels={
        "gdpPercap": "GDP per capita (US$; log scale)",
        "lifeExp": "Life expectancy (years)",
        "pop": "Population",
        "continent_group": "Continent",
    },
    title="Asia spans a wide income range, with life expectancy rising across it",
)

fig.add_annotation(
    x=0.98,
    y=0.98,
    xref="paper",
    yref="paper",
    text="Asia spans a wide income range;<br>higher incomes generally coincide with longer lives.",
    showarrow=False,
    xanchor="right",
    yanchor="top",
    align="right",
    bgcolor="white",
    bordercolor="#2878B5",
    borderwidth=1,
)

fig.update_layout(legend_title_text="Continent")
fig.show()
```
